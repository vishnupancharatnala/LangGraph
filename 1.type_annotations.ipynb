The below are some commonly used type annotations

```python
## TypedDict

from typing import TypedDict

class Human(TypedDict):
    name: str
    age: int
    
person = Human(name = "Alice", age=30)

print(person)

print(person['name'])
```

```python
### Union
# when we want to argument to be one of several types, we can use Union

from typing import Union

def calculate(x: Union[int, str]) -> float:
    return x * x

# a = calculate("q")

    
```

```python
# 3. optional 
# when we want to argument to be either a specific type or None, we can use Optional

from typing import Optional 

def greet(name: Optional[str]) -> str:
    if name is None:
        return "Hello, stranger!"
    else:
        return f"Hello, {name}!"

print(greet("vishnu"))
print(greet(None))
```

```python
# 4. any
# when we want to argument to be of any type, we can use Any
from typing import Any
def process_data(data: Any) -> None:
    print(f"Processing data: {data}")
```

```python
# 5. lambda functions
# when we want to define a small anonymous function, we can use lambda functions
add = lambda x, y: x + y
print(add(3, 5))

nums = [1,2,3]
squared_nums = list(map(lambda x: x**2,nums))
print(squared_nums)
```
